### Área do retângulo

Cria uma função ```area_retangulo(base, altura)``` que recebe dois parâmetros e devolve a área do retângulo. Chama-a com ```base=5``` e ```altura=3```, e exibe o resultado.

````python
def area_retangulo(base, altura):
    return base * altura

area = area_retangulo(5, 3)
print('A área é', area, 'm².')
````

### Menor e maior com parâmetros nomeados

Cria uma função ```min_max(n1, n2)``` que devolve dois valores: o menor e o maior dos dois números recebidos. Chama-a passando os parâmetros por nome, na ordem invertida ```(n2= primeiro, n1= depois)```, e guarda os dois retornos em variáveis separadas.

````python
def min_max(n1, n2):
    resultado = [n1, n2] if n1 < n2 else [n2, n1]
    return resultado

menor, maior = min_max(n2 = 10, n1 = 5)
print(f'O menor é {menor}, e o maior é {maior}')
````

### Média com *args

Cria uma função ```media(*notas)``` que aceita uma quantidade variável de notas e devolve a média delas. Testa com 3 e depois com 4 notas.

````python
def media(*notas):
    return sum(notas) / len(notas)

print('Média das notas 4, 6 e 8:', media(4, 6, 8))
print('Média das notas 16, 16, 16 e 16:', media(16, 16, 16, 16))
````

### Lambda + map(): Celsius para Fahrenheit

Dada ```temperaturas_c = [0, 10, 20, 30, 100]```, usa ```map()``` com uma função lambda para converter todas as temperaturas para Fahrenheit (```F = C * 9/5 + 32```).

````python
converter_c_em_f = lambda c: c * (9 / 5) + 32

temperaturas_c = [0, 10, 20, 30, 100]
temperaturas_f = list(map(converter_c_em_f, temperaturas_c))

print('Temperaturas em ºC:', temperaturas_c)
print('Temperaturas em ºF:', temperaturas_f)
````

### Ficha de cadastro com **kwargs

Cria uma função ```ficha_cadastro(**dados)``` que recebe qualquer quantidade de campos nomeados e devolve uma única string com um campo por linha, no formato "```campo: valor```".

````python
def ficha_cadastro(**dados):
    linhas = []
    for campo, valor in dados.items():
        linhas.append(f'{campo}: {valor}')
    return '\n'.join(linhas)

print(ficha_cadastro(nome = 'Eliandro', apelido = 'Sérgio', idade = 21, altura = 1.70, data_de_nascimento = '2004-10-03'))
````

### Prevê a saída: UnboundLocalError

Prevê a saída deste código e explica o papel do comando global. O que aconteceria se, dentro de ```incrementar_global()```, a linha ```global contador``` fosse removida?

```
contador = 0

def incrementar_global():
    global contador
    contador = contador + 1
    print(contador)

incrementar_global()
incrementar_global()
print(contador)
```

````python
'''
Sem a linha 'global contador' em 'incrementar_global()', teriamos um UnboundLocalError.
Pois estariamos a tentar criar uma nova variável local com o nome 'contador',
porém, como o comando inicial seria 'contador = contador + 1,
e 'contador' local ainda não existe, não teria como adicionar 'contador + 1' a 'contador',
pois, 'contador' ainda não existe.

Estariamos a tentar acessar o valor de uma variável que não existe,
logo, uma variável sem valor atribuido ainda.
'''

contador = 0

def incrementar_global():
    global contador
    contador = contador + 1
    print(contador)

incrementar_global()
incrementar_global()
print(contador)
````

### Uma função como parâmetro

Cria uma função ```melhor_valor(*valores, criterio)``` que recebe uma quantidade variável de números e uma função de critério (por exemplo, as funções built-in max ou min) passada por nome, e devolve o resultado de aplicar esse critério aos valores recebidos.

````python
def melhor_valor(*valores, criterio):
    return criterio(valores)

notas = [14, 9, 18, 12]
print(melhor_valor(*notas, criterio=max))
print(melhor_valor(*notas, criterio=min))
````
