### Básico 002

As sete operações aritméticas

Pede ao utilizador dois números com input() (converte-os para float) e mostra o resultado das sete operações aritméticas do Python: soma, subtração, multiplicação, divisão, divisão inteira, resto e potenciação.

```python
a = float(input("Primeiro número: "))
b = float(input("Segundo número: "))

print('Soma:', a + b)
print('Subtração:', a - b)
print('Multiplicação:', a * b)
print('Divisão:', a / b)
print('Divisão inteira:', a // b)
print('Resto:', a % b)
print('Potenciação:', a ** b)
```

### Intermediário 001

Cálculo dentro da f-string

### Intermediário 002

Prevê a saída: == contra is

Sem executar, diz o que será impresso em cada uma das quatro linhas abaixo e explica porquê. Depois corre o código para confirmar.

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(a == b)
print(a is b)
print(a is c)
print(b == c)
```

### Intermediário 003

Par ou ímpar, sem usar if

Pede um número inteiro ao utilizador. Sem usar if (as condicionais são o assunto do próximo capítulo), guarda o resultado de uma comparação relacional com o operador % numa variável booleana eh_par, e mostra uma frase com f-string que use essa variável diretamente dentro do texto.

```python
numero: int = int(input('Digite um número inteiro: '))
eh_par: bool = numero % 2 == 0
print(f'O número {numero} é par? R.: {eh_par}')
```

## Avançados

### Avançado 001

Área e perímetro com formatação

Uma pizzaria quer calcular a área e o perímetro (circunferência) de pizzas redondas. Pede ao utilizador o raio da pizza em cm e, usando a biblioteca math e f-strings com formatação de duas casas decimais (:.2f), mostra a área (π·r²) e o perímetro (2·π·r).

```python
import math

raio: float = float(input('Digite o raio da pizza em cm: '))

# área da pizza
area = math.pi * math.pow(raio, 2)

# perímetro da pizza
perimetro = 2 * math.pi * raio

# exibindo o resultado
print(f'Uma pizza com um raio de {raio:.2f}cm tem uma área de {area:.2f}cm e um perímetro de {perimetro:.2f}cm.')
```

### Avançado 002

Troco em notas, só com // e %

Um multibanco só tem notas de 2000, 1000, 500 e 100 Kz. Escreve um programa que recebe um valor inteiro em Kz (via input()) e calcula quantas notas de cada valor serão entregues, usando apenas divisão inteira (//) e resto (%) em cadeia — sem usar if nem ciclos.

```python
valor: int = int(input('Valor a levantar (Kz): '))

# número de notas
notas_de_2000: int = valor // 2000
resto = valor % 2000
notas_de_1000: int = resto // 1000
resto = resto % 1000
notas_de_500: int = resto // 500
resto = resto % 500
notas_de_100: int = resto // 100

# exibindo o resultado
print('Notas de 2000Kz:', notas_de_2000)
print('Notas de 1000Kz:', notas_de_1000)
print('Notas de 500Kz:', notas_de_500)
print('Notas de 100Kz:', notas_de_100)
```

### Avançado 003

A mesma frase de três formas

Dado nome = "Eliandro" e linguagem = "Python", escreve a mesma frase “Eliandro está a aprender Python.” de três formas diferentes: (1) concatenação com +; (2) o método .format() com índices posicionais; (3) f-string. No final, usa o operador == para confirmar que as três strings resultantes são idênticas.

```python
nome = 'Eliandro'
linguagem = 'Python'

# formando as frases
frase1 = nome + ' está a aprender ' + linguagem + '.'
frase2 = '{0} está a aprender {1}.'.format(nome, linguagem)
frase3 = f'{nome} está a aprender {linguagem}.'

# exibindo as frases
print(frase1)
print(frase2)
print(frase3)

# verificando se as frases exibem resultados idênticos
print(frase1 == frase2 == frase3)
```
