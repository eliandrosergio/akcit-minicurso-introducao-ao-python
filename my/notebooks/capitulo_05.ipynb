# Manipulação de Arquivos

Modos de abertura (r, w, a), o bloco with, leitura e escrita de texto, contagem de linhas e palavras, e tratamento de erros de arquivo.

## Básicos

### Criar um arquivo de notas

Cria a lista ```notas = ["Matemática: 16", "Português: 14", "Informática: 18"]``` e escreve cada item numa linha separada de um arquivo chamado ```notas.txt```, usando with e o modo "w".

````python
notas = ["Matemática: 16", "Português: 14", "Informática: 18"]

with open('/home/eliandro/Downloads/notas.txt', 'w') as arquivo:
    for item in notas:
        arquivo.write(item + '\n')
````

### Mostrar o conteúdo de um arquivo

Escreve uma função ```mostrar_conteudo(caminho)``` que abre o arquivo em modo de leitura e imprime todo o seu conteúdo, usando with.

````python
def mostrar_conteudo(caminho):
    with open(caminho, 'r') as arquivo:
        conteudo = arquivo.read()
    print(conteudo)

mostrar_conteudo('/home/eliandro/Downloads/notas.txt')
````

### Adicionar sem apagar (append)

Usando o modo "a", adiciona a linha "Inglês: 17" ao final de notas.txt, sem apagar o conteúdo já existente. Depois, mostra o conteúdo atualizado.

````python
with open('/home/eliandro/Downloads/notas.txt', 'a') as arquivo:
    arquivo.write('Inglês: 17\n')

mostrar_conteudo('/home/eliandro/Downloads/notas.txt')
````

## Intermediários

### Contar linhas de um arquivo

Escreve uma função ```contar_linhas(caminho)``` que devolve quantas linhas o arquivo tem, usando ```readlines()``` e ```len()```.

````python
def contar_linhas(caminho):
    with open(caminho, 'r') as arquivo:
        return len(arquivo.readlines())

print(contar_linhas('/home/eliandro/Downloads/notas.txt'))
````

### Contar ocorrências de uma palavra

Escreve uma função ```contar_palavra(caminho, palavra)``` que conta quantas vezes uma palavra aparece no arquivo, usando ```.split()``` e ```.count()```.

````python
def contar_palavra(caminho, palavra):
    with open(caminho, 'r') as arquivo:
        return ((arquivo.read()).split()).count(palavra)

contar_palavra('/home/eliandro/Downloads/notas.txt', 'Matemática:')
````

### Linhas de um arquivo numa lista, sem \n

Escreve uma função ```linhas_para_lista(caminho)``` que lê o arquivo e devolve uma lista com cada linha, mas sem a quebra de linha (\n) no final de cada elemento.

````python
def linhas_para_lista(caminho):
    with open(caminho, 'r') as arquivo:
        linhas = arquivo.readlines()
    return [linha.strip() for linha in linhas]

print(linhas_para_lista('/home/eliandro/Downloads/notas.txt'))
````

### Tratar arquivo inexistente

Escreve uma função ```mostrar_conteudo_seguro(caminho)``` que tenta mostrar o conteúdo de um arquivo e, se ele não existir, mostra uma mensagem amigável em vez de o programa terminar com um erro.

````python
def mostrar_conteudo_seguro(caminho):
    try:
        with open(caminho, 'r') as arquivo:
            conteudo = arquivo.read()
        print(conteudo)
    except FileNotFoundError:
        print(f"O arquivo '{caminho}' não foi encontrado!")

mostrar_conteudo_seguro('/home/eliandro/Downloads/Elian.elian')
mostrar_conteudo_seguro('/home/eliandro/Downloads/notas.txt')
````
