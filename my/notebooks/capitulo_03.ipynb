## Básicos

### Classificar uma nota

Pede uma nota (0 a 20) ao utilizador. Usando if/elif/ else, imprime "Aprovado" se a nota for ≥ 10, "Recuperação" se estiver entre 7 (incluído) e 10 (excluído), e "Reprovado" caso contrário.

````python
nota = float(input('Digite sua nota: '))

if nota >= 10:
    print('Aprovado')
elif nota >= 7 and nota < 10:
    print('Recuperação')
else:
    print('Reprovado')
````

### Tabuada com range()

Pede um número inteiro ao utilizador e usa um for com range() para imprimir a sua tabuada, de 1 a 10 (por exemplo, "7 x 1 = 7").

````python
num = int(input('Digite um número inteiro: '))

for i in range(1, 11):
    print(f'{num} x {i} = {num * i}')
````

### Somar até ultrapassar 100

Usa um while para somar números inteiros consecutivos a partir de 1 (1, depois 1+2, depois 1+2+3…) até a soma ultrapassar 100. No final, mostra quantos números foram somados e qual foi a soma total.

````python
soma: int = 0
num: int = 0

while soma <= 100:
    num += 1
    soma += num

print('Quantidade de números somados:', num)
print('A soma total:', soma)
````

## Intermediários

### Lista numerada com enumerate()

Dada ```ingredientes = ["farinha", "ovos", "açúcar", "manteiga"]```, usa for com enumerate() para imprimir uma lista numerada a partir de 1, no formato "1. farinha".

````python
ingredientes = ["farinha", "ovos", "açúcar", "manteiga"]

for i, ingrediente in enumerate(ingredientes, start = 1):
    print(f'{i}. {ingrediente}')
````

### List comprehension: cubos pares

Usando list comprehension (numa única linha), cria uma lista com os cubos (x**3) de todos os números pares entre 0 e 20, incluindo o 20.

````python
lista_cubos = [i ** 3 for i in range(21) if i % 2 == 0]
print(lista_cubos)
````

### break e continue

Dada ```valores = [4, 8, 15, -3, 16, 23, 42]```:
- (a) usa for + break para percorrer a lista e parar assim que encontrar o primeiro número negativo, mostrando cada número visto até aí;
- (b) num novo for sobre a mesma lista, usa continue para não mostrar os múltiplos de 3.

````python
valores = [4, 8, 15, -3, 16, 23, 42]

for num in valores:
    if num < 0:
        break
    print(num)

print(f'\n{'-' * 20}\n')

for num in valores:
    if num % 3 == 0:
        continue
    print(num)
````

### try-except com erros específicos

Pede dois números ao utilizador e tenta dividir o primeiro pelo segundo, tratando separadamente dois erros possíveis: ValueError (o utilizador digitou algo que não é um número) e ZeroDivisionError (o segundo número é 0), cada um com a sua própria mensagem.

````python
try:
    num1 = float(input('Digite o 1º número: '))
    num2 = float(input('Digite o 2º número: '))
    resultado = num1 / num2
    print('Resultado:', resultado)
except ValueError:
    print('Erro: conteúdo digitado não é um número!')
except ZeroDivisionError:
    print('Erro: divisão por zero não é permitido!')
````

### match-case para códigos HTTP

Pede ao utilizador um código de status HTTP (inteiro) e usa match/case para mostrar a mensagem correspondente:
- 200 → “OK”
- 301 → “Redirecionado”
- 404 → “Não encontrado”
- 500 → “Erro interno do servidor”
- 400 ou 401 juntos → “Pedido inválido ou não autorizado”

Qualquer outro valor deve mostrar “Código desconhecido”.

````python
num = int(input('Código de status HTTP (número inteiro): '))

match num:
    case 200:
        print('OK')
    case 301:
        print('Redirecionado')
    case 404:
        print('Não encontrado')
    case 500:
        print('Erro interno do servidor')
    case 400 | 401:
        print('Pedido inválido ou não autorizado')
    case _:
        print('Código desconhecido')
````

### Dict comprehension com filtro

Dada ```palavras = ["sol", "computador", "python", "eu", "programação", "ia"]```, cria (numa linha, com dict comprehension) um dicionário que associe cada palavra com mais de 3 letras ao seu número de letras.

````python
palavras = ["sol", "computador", "python", "eu", "programação", "ia"]

dict_palavras = {palavra: len(palavra) for palavra in palavras if len(palavra) > 3}
print(dict_palavras)
````
